```python
# Remove duplicate records
df = df.drop_duplicates()

# Remove rows with missing Customer IDs
df = df.dropna(subset=["CustomerID"])

# Convert CustomerID into integer
df["CustomerID"] = df["CustomerID"].astype(int)

# Convert InvoiceDate into datetime format
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Remove cancelled invoices
df = df[~df["InvoiceNo"].astype(str).str.startswith("C")].copy()

# Remove invalid quantities and prices
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)].copy()

# Calculate total transaction amount
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

print("Data cleaning completed!")
print("Cleaned dataset shape:", df.shape)

display(df.head())
```

```python
# Set reference date
reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

# Calculate Recency, Frequency, and Monetary
rfm = df.groupby("CustomerID").agg({
    "InvoiceDate": lambda x: (reference_date - x.max()).days,
    "InvoiceNo": "nunique",
    "TotalPrice": "sum"
})

# Rename columns
rfm.rename(columns={
    "InvoiceDate": "Recency",
    "InvoiceNo": "Frequency",
    "TotalPrice": "Monetary"
}, inplace=True)

print("RFM Analysis Completed!")

display(rfm.head(10))
```

```python
# Calculate Recency score
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Calculate Frequency score
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Calculate Monetary score
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Calculate total RFM score
rfm["RFM_Score"] = (
    rfm["R_Score"] +
    rfm["F_Score"] +
    rfm["M_Score"]
)

print("RFM Scoring Completed!")

display(rfm.head(10))
```

```python
def segment_customer(score):

    if score >= 13:
        return "Champions"

    elif score >= 10:
        return "Loyal Customers"

    elif score >= 7:
        return "Potential Customers"

    elif score >= 5:
        return "At Risk"

    else:
        return "Hibernating"


# Assign customer segments
rfm["Segment"] = rfm["RFM_Score"].apply(segment_customer)

print("Customer Segmentation Completed!")

display(rfm.head(10))
```

```python
segment_summary = rfm.groupby("Segment").agg(
    Recency=("Recency", "mean"),
    Frequency=("Frequency", "mean"),
    Monetary=("Monetary", "mean"),
    Customer_Count=("Segment", "size")
).round(2)

print("Customer Segment Summary:")

display(segment_summary)
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=rfm,
    x="Segment",
    order=rfm["Segment"].value_counts().index
)

plt.title("Customer Distribution by Segment")

plt.xlabel("Customer Segment")

plt.ylabel("Number of Customers")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

heatmap_data = rfm.groupby("Segment")[
    ["Recency", "Frequency", "Monetary"]
].mean()

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)

plt.title("RFM Metrics by Customer Segment")

plt.tight_layout()

plt.show()
```
